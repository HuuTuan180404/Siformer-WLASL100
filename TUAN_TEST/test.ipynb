Dựa trên quá trình phân tích sâu của chúng ta, đây là những hướng cải tiến tiềm năng cho mô hình `SiFormer`, được phân loại theo từng giai đoạn.

### **1. Cải tiến về Dữ liệu (Nền tảng quan trọng nhất)**

Đây thường là nơi mang lại hiệu quả cao nhất. "Rác vào, rác ra".

*   **Tăng cường chất lượng Keypoint:**
    *   **Vấn đề:** Mô hình hoàn toàn phụ thuộc vào chất lượng của các điểm khớp (keypoint) được trích xuất từ video. Nếu bộ trích xuất keypoint (ví dụ: OpenPose, MediaPipe phiên bản cũ) hoạt động kém, mô hình sẽ học trên dữ liệu nhiễu.
    *   **Cải tiến:** Sử dụng một bộ ước tính tư thế (pose estimator) hiện đại và chính xác hơn, ví dụ như **Google MediaPipe Holistic** phiên bản mới nhất. Việc chạy lại toàn bộ dữ liệu qua một bộ trích xuất tốt hơn có thể tạo ra một tập dữ liệu "sạch" hơn đáng kể, giúp mô hình học các mẫu tinh tế tốt hơn.

*   **Tăng cường dữ liệu (Data Augmentation) thông minh hơn:**
    *   **Hiện tại:** Mô hình có các phép tăng cường về không gian (xoay, trượt,...) và thêm nhiễu.
    *   **Cải tiến (Tập trung vào chiều thời gian - Temporal Augmentation):** Ngôn ngữ ký hiệu là một chuỗi động.
        *   **Frame Dropping:** Xóa ngẫu nhiên một vài khung hình trong video. Điều này buộc mô hình phải học cách suy luận từ thông tin không đầy đủ, làm nó mạnh mẽ hơn.
        *   **Temporal Resampling:** Thay đổi tốc độ của video một cách ngẫu nhiên (làm nhanh hơn hoặc chậm hơn một chút). Điều này giúp mô hình không bị phụ thuộc vào một tốc độ ký hiệu cố định.
        *   **Time Masking:** Tương tự như trong mô hình âm thanh, che đi (mask) một đoạn ngắn của chuỗi thời gian (vài khung hình liên tiếp) và buộc mô hình phải dự đoán dựa trên bối cảnh xung quanh.

---

### **2. Cải tiến về Tiền xử lý (Tinh chỉnh "công thức")**

*   **Thử nghiệm với các phương pháp Chuẩn hóa (Normalization) khác:**
    *   **Vấn đề:** Các thuật toán `normalize_body` và `normalize_hand` hiện tại dựa trên một số giả định (ví dụ: Bounding Box hình vuông `6x6 head_metric`).
    *   **Cải tiến:**
        *   **Thay đổi kích thước Bounding Box:** Liệu `6x6` có phải là tối ưu? Một cái hộp `8x8` hoặc một hình chữ nhật `6x8` có thể bao quát được không gian ký hiệu tốt hơn và ít cắt mất các cử chỉ tay ở xa không? Đây là một siêu tham số có thể thử nghiệm.
        *   **Thêm đặc trưng về chiều sâu (Z-coordinate):** Nếu bộ trích xuất keypoint cung cấp tọa độ Z (khoảng cách tương đối tới camera), việc thêm nó vào như một chiều đặc trưng thứ ba có thể cung cấp cho mô hình thông tin 3D vô giá.

*   **Kỹ thuật đặc trưng (Feature Engineering):**
    *   **Hiện tại:** Đầu vào chỉ là tọa độ (vị trí).
    *   **Cải tiến:** Tính toán và thêm các đặc trưng động học.
        *   **Vận tốc:** `vị_trí(t) - vị_trí(t-1)`.
        *   **Gia tốc:** `vận_tốc(t) - vận_tốc(t-1)`.
        *   Việc nối các vector vận tốc và gia tốc vào vector tọa độ ban đầu sẽ cung cấp cho mô hình thông tin rõ ràng về "chuyển động" và "sự thay đổi chuyển động", thay vì buộc nó phải tự học điều này từ chuỗi các vị trí.

---

### **3. Cải tiến về Kiến trúc Mô hình (Thay đổi "bộ não")**

Đây là nơi thú vị nhất.

*   **Cơ chế Kết hợp Đặc trưng (Fusion Mechanism) tốt hơn:**
    *   **Hiện tại:** Sau khi qua 3 encoder riêng biệt, các "bản báo cáo" (`memory`) được nối lại một cách đơn giản bằng `torch.cat`.
    *   **Cải tiến:**
        *   **Cross-Attention Fusion:** Đây là một ý tưởng rất mạnh mẽ. Thay vì chỉ nối lại, hãy để các luồng "giao tiếp" với nhau. Ví dụ: tạo một tầng mà ở đó, `l_hand_memory` sẽ đóng vai trò là Query, còn `body_memory` và `r_hand_memory` đóng vai trò là Key và Value. Điều này cho phép tay trái "hỏi" rằng: "Khi tôi đang ở hình dạng này, thì tay phải và cơ thể đang làm gì?". Nó tạo ra một sự kết hợp ngữ cảnh sâu sắc hơn nhiều.
        *   **Gated Fusion:** Thêm một cơ chế "cổng" (gating mechanism) nhỏ. Đây là một mạng nơ-ron nhỏ học cách tạo ra các trọng số (ví dụ: `[w1, w2, w3]`) cho mỗi khung hình. `full_memory` sẽ là `w1*l_hand + w2*r_hand + w3*body`. Điều này cho phép mô hình tự động quyết định xem tại một thời điểm nhất định, thông tin từ tay trái, tay phải hay cơ thể là quan trọng hơn.

*   **Cải thiện mô hình hóa thời gian (Temporal Modeling):**
    *   **Hiện tại:** Sử dụng Positional Encoding có thể học được.
    *   **Cải tiến:**
        *   **Convolutional Stem:** Thêm một vài lớp tích chập 1D (Conv1D) ngay sau bước embedding. Các lớp Conv1D rất giỏi trong việc nắm bắt các mẫu cục bộ (ví dụ: chuyển động của ngón tay trong 3-5 khung hình liên tiếp). Đầu ra của các lớp Conv này sau đó sẽ được đưa vào các encoder Transformer, vốn rất giỏi trong việc nắm bắt các mối quan hệ toàn cục (cả câu). Sự kết hợp này (local + global) thường rất hiệu quả.
        *   **Relative Positional Encoding:** Thay vì mã hóa vị trí tuyệt đối, hãy mã hóa khoảng cách tương đối giữa các khung hình. Điều này có thể giúp mô hình tổng quát hóa tốt hơn với các video có độ dài khác nhau.

---

### **4. Cải tiến về Quy trình Huấn luyện**

*   **Hàm mất mát (Loss Function):**
    *   **Hiện tại:** `CrossEntropyLoss` tiêu chuẩn.
    *   **Cải tiến:**
        *   **Label Smoothing:** Thay vì yêu cầu mô hình dự đoán xác suất là `[0, 0, 1, 0, ...]`, chúng ta yêu cầu nó dự đoán một cái gì đó "mềm" hơn, ví dụ `[0.01, 0.01, 0.97, 0.01, ...]`. Kỹ thuật này giúp mô hình bớt "tự tin thái quá" và thường cải thiện khả năng tổng quát hóa.
        *   **Focal Loss:** Nếu tập dữ liệu của bạn bị mất cân bằng (một số ký hiệu xuất hiện nhiều hơn hẳn các ký hiệu khác), Focal Loss là một lựa chọn tuyệt vời. Nó sẽ tập trung vào việc học các mẫu khó, ít xuất hiện hơn.

*   **Bộ lập lịch Tốc độ học (Scheduler):**
    *   **Hiện tại:** `MultiStepLR`.
    *   **Cải tiến:** Thử nghiệm với các scheduler mượt mà hơn như `CosineAnnealingLR`. Nó giảm tốc độ học một cách từ từ theo đường cong cosine, thường giúp mô hình hội tụ đến một điểm tối ưu tốt hơn.

Đây là những hướng đi khả thi. Mỗi ý tưởng cải tiến này đều có thể là một đề tài thí nghiệm nhỏ, nơi bạn đưa ra một giả thuyết, thay đổi mã nguồn, huấn luyện lại mô hình và so sánh kết quả để xem liệu sự thay đổi đó có thực sự mang lại hiệu quả hay không.

Chắc chắn rồi! Sử dụng `CosineAnnealingLR` là một lựa chọn rất phổ biến và thường mang lại hiệu quả tốt. Bộ lập lịch này sẽ điều chỉnh tốc độ học (learning rate) theo một đường cong cosine, giảm từ từ giá trị ban đầu xuống một giá trị tối thiểu trong một khoảng thời gian nhất định.

Để thay thế `MultiStepLR` bằng `CosineAnnealingLR`, bạn chỉ cần thay đổi dòng khởi tạo scheduler.

### **Cách thực hiện**

Bạn sẽ thay thế dòng mã này:

**Mã gốc:**
```python
scheduler = optim.lr_scheduler.MultiStepLR(optimizer, milestones=[60, 80], gamma=0.1)  # 40, 60, 80  Đây là bộ lập lịch tốc độ học (learning rate scheduler). Nó sẽ tự động giảm tốc độ học tại các epoch nhất định (milestones=[60, 80]) để giúp mô hình hội tụ tốt hơn.
```

**Bằng dòng mã mới này:**
```python
# Đây là bộ lập lịch tốc độ học (learning rate scheduler). Nó sẽ tự động điều chỉnh tốc độ học theo đường cong cosine.
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=args.epochs, eta_min=0) 
```

---

### **Giải thích chi tiết**

1.  **`torch.optim.lr_scheduler.CosineAnnealingLR`**: Đây là lớp scheduler bạn muốn sử dụng.

2.  **`optimizer`**: Tham số đầu tiên vẫn là bộ tối ưu hóa (optimizer) mà bạn đã khởi tạo (`torch.optim.AdamW`). Scheduler sẽ điều chỉnh tốc độ học của optimizer này.

3.  **`T_max`**: Đây là tham số quan trọng nhất.
    *   Nó xác định số lượng epoch (hoặc vòng lặp) để hoàn thành một nửa chu kỳ của đường cong cosine, tức là khoảng thời gian để tốc độ học giảm từ giá trị cao nhất (ban đầu) xuống giá trị thấp nhất (`eta_min`).
    *   Một cách làm rất phổ biến và hiệu quả là đặt `T_max` bằng tổng số epoch bạn dự định huấn luyện (`args.epochs`). Điều này có nghĩa là tốc độ học sẽ giảm một cách mượt mà trong suốt quá trình huấn luyện, bắt đầu từ giá trị ban đầu tại epoch 0 và đạt đến giá trị tối thiểu ở epoch cuối cùng.

4.  **`eta_min`** (tùy chọn):
    *   Đây là giá trị tốc độ học tối thiểu mà scheduler có thể đạt tới.
    *   Mặc định của nó là `0`. Đặt `eta_min=0` có nghĩa là vào cuối chu kỳ `T_max`, tốc độ học sẽ giảm xuống 0. Đây là một giá trị mặc định an toàn và hiệu quả.

### **Tại sao nên dùng `CosineAnnealingLR`?**

*   **Giảm mượt mà**: Không giống như `MultiStepLR` giảm tốc độ học một cách đột ngột tại các `milestones` cụ thể, `CosineAnnealingLR` giảm nó một cách từ từ. Điều này giúp mô hình ổn định và hội tụ vào một điểm tối ưu tốt hơn mà không "nhảy" qua nó do sự thay đổi đột ngột của tốc độ học.
*   **Ít siêu tham số hơn**: Bạn không cần phải đoán xem nên đặt các `milestones` ở đâu. Bạn chỉ cần xác định tổng thời gian huấn luyện (`T_max`), điều này đơn giản hơn nhiều.
*   **Hiệu quả đã được chứng minh**: Trong nhiều bài toán thị giác máy tính và học sâu, `CosineAnnealingLR` thường cho kết quả tốt hơn hoặc tương đương so với các scheduler khác.

Bạn không cần thay đổi bất cứ điều gì khác trong vòng lặp huấn luyện. Cuộc gọi `scheduler.step()` (có thể nằm trong hàm `train_epoch` của bạn) sẽ hoạt động đúng với scheduler mới này.

Đầu tiên, bạn cần một module TCN. TCN không phải là một lớp có sẵn trong PyTorch, vì vậy chúng ta cần tự định nghĩa nó.

import torch
import torch.nn as nn
from torch.nn.utils import weight_norm

class TemporalBlock(nn.Module):
    def __init__(self, n_inputs, n_outputs, kernel_size, stride, dilation, padding, dropout=0.2):
        super(TemporalBlock, self).__init__()
        self.conv1 = weight_norm(nn.Conv1d(n_inputs, n_outputs, kernel_size,
                                           stride=stride, padding=padding, dilation=dilation))
        self.chomp1 = nn.Identity()  # Sẽ được thay thế để cắt bỏ padding

        self.relu1 = nn.ReLU()
        self.dropout1 = nn.Dropout(dropout)

        self.conv2 = weight_norm(nn.Conv1d(n_outputs, n_outputs, kernel_size,
                                           stride=stride, padding=padding, dilation=dilation))
        self.chomp2 = nn.Identity() # Sẽ được thay thế để cắt bỏ padding

        self.relu2 = nn.ReLU()
        self.dropout2 = nn.Dropout(dropout)

        self.net = nn.Sequential(self.conv1, self.chomp1, self.relu1, self.dropout1,
                                 self.conv2, self.chomp2, self.relu2, self.dropout2)
        
        # Lớp Conv1d 1x1 để đảm bảo kích thước của kết nối dư (residual connection) khớp nhau
        self.downsample = nn.Conv1d(n_inputs, n_outputs, 1) if n_inputs != n_outputs else None
        self.relu = nn.ReLU()
        self.init_weights()

    def init_weights(self):
        self.conv1.weight.data.normal_(0, 0.01)
        self.conv2.weight.data.normal_(0, 0.01)
        if self.downsample is not None:
            self.downsample.weight.data.normal_(0, 0.01)

    def forward(self, x):
        # Cắt bỏ phần padding thừa ở phía bên phải để đảm bảo tính nhân quả (causal)
        # Đầu ra không được nhìn thấy tương lai
        out = self.net[0](x)
        out = out[:, :, :-self.net[0].padding[0]]
        out = self.net[1:](out)
        
        res = x if self.downsample is None else self.downsample(x)
        return self.relu(out + res)


class TemporalConvNet(nn.Module):
    def __init__(self, num_inputs, num_channels, kernel_size=2, dropout=0.2):
        """
        Khởi tạo Mạng Tích chập Thời gian.
        Args:
            num_inputs (int): Số chiều đặc trưng đầu vào (num_feats).
            num_channels (list): Danh sách số kênh ẩn trong mỗi lớp.
            kernel_size (int): Kích thước của kernel tích chập.
            dropout (float): Tỷ lệ dropout.
        """
        super(TemporalConvNet, self).__init__()
        layers = []
        num_levels = len(num_channels)
        for i in range(num_levels):
            dilation_size = 2 ** i
            in_channels = num_inputs if i == 0 else num_channels[i-1]
            out_channels = num_channels[i]
            layers += [TemporalBlock(in_channels, out_channels, kernel_size, stride=1, dilation=dilation_size,
                                     padding=(kernel_size-1) * dilation_size, dropout=dropout)]

        self.network = nn.ModuleList(layers)

    def forward(self, x):
        """
        Input x có dạng (N, C, L)
            N: batch size
            C: số kênh/đặc trưng (num_feats)
            L: độ dài chuỗi (seq_len)
        """
        for layer in self.network:
            x = layer(x)
        return x

            self.input_projection = nn.Identity()
        else:
            # Giữ lại hành vi gốc nếu không sử dụng TCN

### Bối cảnh và Vấn đề

Các mô hình Transformer truyền thống trong nhận dạng hành động thường hoạt động theo cách sau:
1.  Lấy **tất cả** các keypoint của cơ thể (mũi, mắt, vai, khuỷu tay, cổ tay, các khớp ngón tay...) trong một frame.
2.  Coi chúng như một chuỗi các "từ" (tokens).
3.  Ném toàn bộ chuỗi này vào một bộ mã hóa (Encoder) duy nhất.
4.  Bộ mã hóa sẽ dùng cơ chế tự chú ý (self-attention) để tìm ra mối quan hệ giữa **mọi keypoint với mọi keypoint khác**.

**Vấn đề của cách tiếp cận này là gì?**

1.  **Chi phí tính toán khổng lồ:** Cơ chế tự chú ý có độ phức tạp là `O(n^2)`, với `n` là độ dài chuỗi. Nếu bạn có nhiều keypoint, việc tính toán mối quan hệ giữa "mọi thứ với mọi thứ" sẽ rất chậm và tốn tài nguyên.
2.  **Mối quan hệ không cần thiết:** Chuyển động của khớp ngón tay cái bên tay trái có thực sự liên quan mật thiết đến chuyển động của mắt phải không? Có lẽ là không. Việc bắt mô hình phải học các mối quan hệ xa xôi và không liên quan này có thể gây nhiễu và làm giảm hiệu quả.
3.  **Dễ bị ảnh hưởng bởi dữ liệu thiếu:** Nếu bộ trích xuất tư thế không nhận diện được bàn tay trái trong một vài frame (một vấn đề rất phổ biến), toàn bộ chuỗi đầu vào sẽ bị ảnh hưởng, và thông tin "nhiễu" này sẽ lan truyền ra toàn bộ mô hình trong quá trình tự chú ý, làm ảnh hưởng đến cả việc nhận diện chuyển động của cơ thể và tay phải.

### Giải pháp của Siformer: "Chia để trị"

Thay vì xử lý tất cả trong một, Siformer áp dụng triết lý "chia để trị" bằng **Kiến trúc Tách biệt Đặc trưng**.

**Cách hoạt động (Xem Hình 3 trong bài báo):**

1.  **Phân tách dữ liệu đầu vào:** Thay vì gộp chung, dữ liệu keypoint được chia thành **3 nhóm riêng biệt**:
    *   **Nhóm 1:** Tất cả các keypoint của **tay trái**.
    *   **Nhóm 2:** Tất cả các keypoint của **tay phải**.
    *   **Nhóm 3:** Tất cả các keypoint của **phần thân trên** (mũi, cổ, vai, khuỷu tay, cổ tay).

2.  **Ba Luồng Mã hóa Song song:** Mỗi nhóm dữ liệu trên được đưa vào một **bộ mã hóa Transformer (Encoder) độc lập**.
    *   **Encoder Tay Trái:** Chỉ tập trung học mối quan hệ không gian-thời gian **giữa các khớp của tay trái**. Nó chỉ trả lời câu hỏi: "Các ngón tay và cổ tay trái đang chuyển động như thế nào?"
    *   **Encoder Tay Phải:** Tương tự, chỉ tập trung vào tay phải.
    *   **Encoder Thân:** Chỉ tập trung học chuyển động của phần thân trên.

3.  **Kết hợp (Concatenation):** Sau khi mỗi Encoder đã xử lý xong và tạo ra một "bản đồ đặc trưng" (feature map) đã được mã hóa cho từng bộ phận, ba bản đồ đặc trưng này sẽ được **nối (concatenate)** lại với nhau để tạo thành một bản đồ đặc trưng tổng hợp.
    *   `F_map = concat(F_left_hand, F_right_hand, F_body)` (Công thức (5) trong bài báo)

4.  **Giải mã (Decoding):** Bản đồ đặc trưng tổng hợp này sau đó được đưa vào bộ giải mã (Decoder) để đưa ra dự đoán cuối cùng về ký hiệu đang được thực hiện.

### Các thành phần phụ trong Phần 3.3

*   **Learnable Frame-wised Positional Encoding (Mã hóa Vị trí theo Frame có thể học được):**
    *   Transformer vốn không phân biệt thứ tự. Để nó hiểu được chuỗi thời gian, ta cần thêm thông tin vị trí.
    *   Cách làm của Siformer rất thông minh: Tất cả các keypoint **trong cùng một frame** sẽ có cùng một mã hóa vị trí. Các keypoint ở các frame khác nhau sẽ có mã hóa vị trí khác nhau.
    *   Mã hóa này không phải là cố định (như hàm sin/cos) mà là các giá trị được **học** trong quá trình training, cho phép mô hình tự tìm ra cách biểu diễn thứ tự thời gian tốt nhất.

*   **ProbSparse Self-Attention Mechanism:**
    *   Đây là một phiên bản cải tiến của cơ chế tự chú ý, giúp giảm độ phức tạp từ `O(n^2)` xuống `O(n * log n)`.
    *   Thay vì tính toán sự tương quan của một "query" với tất cả các "key" khác, nó chỉ chọn ra một vài "key" có khả năng liên quan cao nhất để tính toán. Điều này giúp tăng tốc độ đáng kể mà không làm giảm nhiều hiệu suất.

### Lợi ích của toàn bộ kiến trúc này

1.  **Chống chịu lỗi (Robustness) tốt hơn:** Nếu dữ liệu tay trái bị mất, chỉ có Encoder tay trái bị ảnh hưởng. Hai Encoder còn lại vẫn hoạt động bình thường, và mô hình vẫn có thể đưa ra dự đoán chính xác dựa trên chuyển động của tay phải và cơ thể.
2.  **Hiệu quả tính toán cao hơn:** Xử lý 3 chuỗi ngắn song song sẽ nhanh hơn nhiều so với xử lý 1 chuỗi dài gộp lại.
3.  **Học đặc trưng tập trung hơn:** Mỗi Encoder chỉ tập trung vào một nhiệm vụ duy nhất (ví dụ: học chuyển động ngón tay), giúp nó học được các đặc trưng cục bộ (local features) một cách tinh vi và chính xác hơn.

**Tóm lại, Phần 3.3 mô tả một kiến trúc Transformer thông minh, nơi các bộ phận cơ thể được phân tích độc lập trong các luồng song song trước khi được kết hợp lại. Cách tiếp cận này giúp mô hình trở nên nhanh hơn, mạnh mẽ hơn trước dữ liệu thiếu, và học được các đặc trưng một cách hiệu quả hơn.**
